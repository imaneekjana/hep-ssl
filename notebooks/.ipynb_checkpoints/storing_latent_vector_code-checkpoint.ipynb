```python
"""
Loading the trained model & stat
"""


"""
Loading trained Encoder
"""

#model = PointNetEncoder()

#folder = '/global/cfs/cdirs/m4474/aneek/particlemind_aneek/saved_models_colliderml_latest/pointnet_models_ttbar_dihiggs_ggf_mixed_events_no_edges/'


 


model = GravNetEncoder()

split = 'dihiggs'
rot = 2*np.pi
energy_noise = 0.05

folder = '/global/cfs/cdirs/m4474/aneek/particlemind_aneek/saved_models_colliderml_latest/gravnet_models_{}_rot_{:.2f}_noise_{:.2f}/'.format(split, rot, energy_noise)
   

#folder = '/global/cfs/cdirs/m4474/aneek/particlemind_aneek/saved_models_colliderml_latest/gravnet_models_ttbar_dihiggs_ggf_mixed_events_no_edges/'


prev_weights = torch.load(folder+'model_epoch_20.pth', map_location='cpu')   # add this)

model.load_state_dict(prev_weights)

model.projection_head = nn.Identity()
model.head = nn.Identity()

stats = np.load(folder + 'stats.npy', allow_pickle=True).item()

MEANS = stats['means']

STDS = stats['stds']

folder = folder + 'without_projection_'
```

```python
"""
Creating original and augmented dataset
"""

augment = Compose([
    RandomRotateXY((0, 2*np.pi))
    ,
    RandomShift((100.0, 100.0, 100.0)),
    RandomSpatialCrop(0.2),
    EnergyWhiteNoise(0.05)
])

dataset_test_ttbar_og = ContrastiveLearningGraphDataset(ContrastiveLearningDataset(dataset_test_ttbar, MEANS, STDS, transform=None), builder=EventGraphBuilder(radius=0))
dataset_test_dihiggs_og = ContrastiveLearningGraphDataset(ContrastiveLearningDataset(dataset_test_dihiggs, MEANS, STDS, transform=None), builder=EventGraphBuilder(radius=0))
dataset_test_ggf_og = ContrastiveLearningGraphDataset(ContrastiveLearningDataset(dataset_test_ggf, MEANS, STDS, transform=None), builder=EventGraphBuilder(radius=0))

loader_test_ttbar_og = DataLoader(dataset_test_ttbar_og, batch_size=16, shuffle=False)
loader_test_dihiggs_og = DataLoader(dataset_test_dihiggs_og, batch_size=16, shuffle=False)
loader_test_ggf_og = DataLoader(dataset_test_ggf_og, batch_size=16, shuffle=False)


dataset_test_ttbar_aug = ContrastiveLearningGraphDataset(ContrastiveLearningDataset(dataset_test_ttbar, MEANS, STDS, transform=augment), builder=EventGraphBuilder(radius=0))
dataset_test_dihiggs_aug = ContrastiveLearningGraphDataset(ContrastiveLearningDataset(dataset_test_dihiggs, MEANS, STDS, transform=augment), builder=EventGraphBuilder(radius=0))
dataset_test_ggf_aug = ContrastiveLearningGraphDataset(ContrastiveLearningDataset(dataset_test_ggf, MEANS, STDS, transform=augment), builder=EventGraphBuilder(radius=0))

loader_test_ttbar_aug = DataLoader(dataset_test_ttbar_aug, batch_size=16, shuffle=False)
loader_test_dihiggs_aug = DataLoader(dataset_test_dihiggs_aug, batch_size=16, shuffle=False)
loader_test_ggf_aug = DataLoader(dataset_test_ggf_aug, batch_size=16, shuffle=False)

device = 'cuda'
model = model.to(device)
model.eval()

latents_ttbar_og = []
latents_dihiggs_og = []
latents_ggf_og = []

latents_ttbar_aug = []
latents_dihiggs_aug = []
latents_ggf_aug = []

"""
storing latents of originals
"""

with torch.no_grad():

    for graph, _ in tqdm(loader_test_ttbar_og):

        graph = graph.to(device)
        latents = model(graph)
        latents_ttbar_og.append(latents.detach())

    for graph, _ in tqdm(loader_test_dihiggs_og):

        graph = graph.to(device)
        latents = model(graph)
        latents_dihiggs_og.append(latents.detach())

    for graph, _ in tqdm(loader_test_ggf_og):

        graph = graph.to(device)
        latents = model(graph)
        latents_ggf_og.append(latents.detach())

    latents_ttbar_og = torch.cat(latents_ttbar_og, dim=0).cpu().numpy()
    latents_dihiggs_og = torch.cat(latents_dihiggs_og, dim=0).cpu().numpy()
    latents_ggf_og = torch.cat(latents_ggf_og, dim=0).cpu().numpy()

    np.savetxt(folder+'ttbar_original_900-1000_latents_epoch_18.txt', latents_ttbar_og)
    np.savetxt(folder+'dihiggs_original_900-1000_latents_epoch_18.txt', latents_dihiggs_og)
    np.savetxt(folder+'ggf_original_900-1000_latents_epoch_18.txt', latents_ggf_og)


"""
storing latents of augmented
"""



with torch.no_grad():

    for graph, _ in tqdm(loader_test_ttbar_aug):

        graph = graph.to(device)
        latents = model(graph)
        latents_ttbar_aug.append(latents.detach())

    for graph, _ in tqdm(loader_test_dihiggs_aug):

        graph = graph.to(device)
        latents = model(graph)
        latents_dihiggs_aug.append(latents.detach())

    for graph, _ in tqdm(loader_test_ggf_aug):

        graph = graph.to(device)
        latents = model(graph)
        latents_ggf_aug.append(latents.detach())

    latents_ttbar_aug = torch.cat(latents_ttbar_aug, dim=0).cpu().numpy()
    latents_dihiggs_aug = torch.cat(latents_dihiggs_aug, dim=0).cpu().numpy()
    latents_ggf_aug = torch.cat(latents_ggf_aug, dim=0).cpu().numpy()

    np.savetxt(folder+'ttbar_augmented_900-1000_latents_epoch_18.txt', latents_ttbar_aug)
    np.savetxt(folder+'dihiggs_augmented_900-1000_latents_epoch_18.txt', latents_dihiggs_aug)
    np.savetxt(folder+'ggf_augmented_900-1000_latents_epoch_18.txt', latents_ggf_aug)




```
